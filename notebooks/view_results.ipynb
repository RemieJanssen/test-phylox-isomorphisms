```python
import pandas as pd
import seaborn as sns
```

```python
path = "../results/networks/200_200/output_no_root_check.csv"
df_no_root = pd.read_csv(path)
path = "./networks/200_200/output.csv"
df = pd.read_csv(path)
df["time_mu_no_root"] = df_no_root["time_mu"]
print(df.describe())


df = df.melt(["n", "k", "id1", "id2"])
df["same_network"] = (df["id1"] == df["id2"])

print(df.describe())


ax = sns.boxplot(data=df, x="variable", y="value", hue="same_network")
ax.set_yscale('log')
```
